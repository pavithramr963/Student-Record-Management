```python
# ============================================================
# PROJECT: STUDENT RECORD MANAGEMENT SYSTEM
# PLATFORM: GOOGLE COLAB
# LANGUAGE: PYTHON
# DATABASE: SQLITE3
# INTERFACE: IPYWIDGETS
# ============================================================

# SPECIFICATIONS:
# 1. Add Student - Add new student details.
# 2. View Students - Display all student records.
# 3. Search Student - Search using Student ID.
# 4. Update Mark - Modify an existing student's mark.
# 5. Delete Student - Delete a student record.
# 6. Clear - Clear all input fields.
# 7. Database - Store records permanently in SQLite
#    within the current Colab runtime.
# 8. Validation - Check ID, name and mark values.

# ============================================================
# STEP 1: IMPORT LIBRARIES
# ============================================================

import sqlite3
import pandas as pd
import ipywidgets as widgets

from IPython.display import display, clear_output

# ============================================================
# STEP 2: DATABASE CONNECTION
# ============================================================

con = sqlite3.connect("student.db")
cur = con.cursor()

cur.execute("""
CREATE TABLE IF NOT EXISTS students (
    id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    mark INTEGER NOT NULL
)
""")

con.commit()

# ============================================================
# STEP 3: CREATE INPUT FIELDS
# ============================================================

student_id = widgets.IntText(
    description="Student ID:",
    value=0
)

student_name = widgets.Text(
    description="Name:",
    placeholder="Enter student name"
)

student_mark = widgets.IntText(
    description="Mark:",
    value=0
)

# ============================================================
# STEP 4: CREATE BUTTONS
# ============================================================

add_btn = widgets.Button(
    description="Add Student",
    button_style="success"
)

view_btn = widgets.Button(
    description="View Students",
    button_style="info"
)

search_btn = widgets.Button(
    description="Search Student",
    button_style="primary"
)

update_btn = widgets.Button(
    description="Update Mark",
    button_style="warning"
)

delete_btn = widgets.Button(
    description="Delete Student",
    button_style="danger"
)

clear_btn = widgets.Button(
    description="Clear"
)

# ============================================================
# STEP 5: CREATE OUTPUT AREA
# ============================================================

output = widgets.Output()

# ============================================================
# STEP 6: CLEAR INPUT FIELDS
# ============================================================

def clear_fields():

    student_id.value = 0
    student_name.value = ""
    student_mark.value = 0

    with output:
        clear_output()

# ============================================================
# STEP 7: ADD STUDENT FUNCTION
# ============================================================

def add_student(b):

    sid = student_id.value
    name = student_name.value.strip()
    mark = student_mark.value

    if sid <= 0 or not name:

        with output:
            clear_output()
            print("Enter valid Student ID and Name")

        return

    if mark < 0 or mark > 100:

        with output:
            clear_output()
            print("Mark must be between 0 and 100")

        return

    try:

        cur.execute(
            "INSERT INTO students VALUES (?, ?, ?)",
            (sid, name, mark)
        )

        con.commit()

        with output:
            clear_output()
            print("Student added successfully!")

        clear_fields()

    except sqlite3.IntegrityError:

        with output:
            clear_output()
            print("Student ID already exists!")

# ============================================================
# STEP 8: VIEW STUDENTS FUNCTION
# ============================================================

def view_students(b):

    with output:
        clear_output()

        cur.execute("SELECT * FROM students")

        records = cur.fetchall()

        if records:

            df = pd.DataFrame(
                records,
                columns=["Student ID", "Name", "Mark"]
            )

            display(df)

        else:

            print("No student records found!")

# ============================================================
# STEP 9: SEARCH STUDENT FUNCTION
# ============================================================

def search_student(b):

    sid = student_id.value

    if sid <= 0:

        with output:
            clear_output()
            print("Enter a valid Student ID")

        return

    cur.execute(
        "SELECT * FROM students WHERE id = ?",
        (sid,)
    )

    record = cur.fetchone()

    with output:
        clear_output()

        if record:

            print("Student Found!")

            student_id.value = record[0]
            student_name.value = record[1]
            student_mark.value = record[2]

            df = pd.DataFrame(
                [record],
                columns=["Student ID", "Name", "Mark"]
            )

            display(df)

        else:

            print("Student not found!")

# ============================================================
# STEP 10: UPDATE MARK FUNCTION
# ============================================================

def update_student(b):

    sid = student_id.value
    mark = student_mark.value

    if sid <= 0:

        with output:
            clear_output()
            print("Enter a valid Student ID")

        return

    if mark < 0 or mark > 100:

        with output:
            clear_output()
            print("Mark must be between 0 and 100")

        return

    cur.execute(
        "UPDATE students SET mark = ? WHERE id = ?",
        (mark, sid)
    )

    con.commit()

    with output:
        clear_output()

        if cur.rowcount > 0:

            print("Student mark updated successfully!")

        else:

            print("Student not found!")

# ============================================================
# STEP 11: DELETE STUDENT FUNCTION
# ============================================================

def delete_student(b):

    sid = student_id.value

    if sid <= 0:

        with output:
            clear_output()
            print("Enter a valid Student ID")

        return

    cur.execute(
        "DELETE FROM students WHERE id = ?",
        (sid,)
    )

    con.commit()

    with output:
        clear_output()

        if cur.rowcount > 0:

            print("Student deleted successfully!")

        else:

            print("Student not found!")

# ============================================================
# STEP 12: CONNECT BUTTONS TO FUNCTIONS
# ============================================================

add_btn.on_click(add_student)

view_btn.on_click(view_students)

search_btn.on_click(search_student)

update_btn.on_click(update_student)

delete_btn.on_click(delete_student)

clear_btn.on_click(lambda b: clear_fields())

# ============================================================
# STEP 13: DESIGN THE APPLICATION INTERFACE
# ============================================================

title = widgets.HTML(
    value="""
    <h2 style='color:#1F4E79;text-align:center;'>
    STUDENT RECORD MANAGEMENT SYSTEM
    </h2>
    """
)

form = widgets.VBox([
    student_id,
    student_name,
    student_mark
])

button_row1 = widgets.HBox([
    add_btn,
    view_btn,
    search_btn
])

button_row2 = widgets.HBox([
    update_btn,
    delete_btn,
    clear_btn
])

app = widgets.VBox([
    title,
    form,
    button_row1,
    button_row2,
    output
])

# ============================================================
# STEP 14: DISPLAY APPLICATION
# ============================================================

display(app)

print("Student Record Management System is Ready!")
```
